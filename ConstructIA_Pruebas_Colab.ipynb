## 3. Prueba adicional de comportamiento combinado
Este caso permite comprobar que se da prioridad a una solicitud de equipo en lugar de responder solamente al saludo.


```python
from constructia_chatbot import ConstructIA
bot = ConstructIA()
for texto in ['Hola, necesito rentar una retroexcavadora', 'Monterrey', '20 de noviembre de 2026', '3 días', 'Con operador', 'Cliente Demo / demo@example.com']:
    respuesta = bot.responder(texto)
    print('Cliente:', texto)
    print('Intención/paso:', bot.ultima_intencion)
    print('ConstructIA:', respuesta)
    print()
```
